```python
# Построения гистограммы распределения

# 1. Настройка стиля и размера поля графика
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid") # светлая тема с сеткой

# 2. Вычисление метрик для вертикальных линий
mean_price = df_clean['Цена'].mean()
median_price = df_clean['Цена'].median()
mode_price = df_clean['Цена'].mode()[0] # находим самую первую моду(самую частую цену)

# 3. Построение гистограммы и линии плотности (KDE)
ax = sns.histplot(
    data=df_clean,
    x='Цена',
    kde=True,           # Добавляем плавную линию плотности
    color='#2b997e',    # Основной цвет (зеленовато-белюзовый(наверное))
    edgecolor='white',  # Белая граница между столбцами для четкости
    bins=35,            # Количество столбцов (интервалов)
    alpha=0.6           # Прозрачность столбцов
)

# 4. Настройка цвета линии KDE
ax.lines[0].set_color('#19694f')
ax.lines[0].set_linewidth(2)

# 5. Добавление пунктирных линий для ключевых метрик
plt.axvline(mean_price, color='#e63946', linestyle='--', linewidth=2, label=f'Среднее: {mean_price:,.0f} руб.')
plt.axvline(median_price, color='#2a649d', linestyle='--', linewidth=2, label=f'Медиана: {median_price:,.0f} руб.')
plt.axvline(mode_price, color='#e9c46a', linestyle='--', linewidth=2, label=f'Мода: {mode_price:,.0f} руб.')

# 6. Оформление заголовков и осей
plt.title('Распределение цен на вилочные погрузчики', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Цена (млн руб.)', fontsize=11)
plt.ylabel('Количество объявлений (шт.)', fontsize=11)

# Форматируем ось X, чтобы цены выводились понятно (например, 2.0M или чистые миллионы)
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"{x/1e3:.1f}M"))

plt.tight_layout()
plt.show()
```

```python
# 1. Создаем отдельную фигуру
plt.figure(figsize=(12, 4))

# 2. Построение Boxplot
ax_box = sns.boxplot(
    data=df_clean,
    x='Цена',
    color='#2b5c8f',          # Тот же основной цвет, что и у гистограммы
    fliersize=5,              # Размер точек-выбросов
    flierprops={              # Настройка внешнего вида выбросов
        'marker': 'o',
        'markerfacecolor': '#e63946', # Точки выбросов выделяем красным
        'alpha': 0.5
    }
)

# 3. Оформление заголовков и осей
plt.title('Диаграмма размаха (Boxplot) цен на погрузчики', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Цена (млн руб.)', fontsize=11)

# Форматируем ось X в миллионы для единого стиля с первым графиком
ax_box.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"{x/1e6:.1f}M"))

plt.tight_layout()
plt.show()
```
